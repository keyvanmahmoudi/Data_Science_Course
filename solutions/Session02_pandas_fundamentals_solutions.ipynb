### آماده‌سازی

کتابخانه‌ها و داده‌هایی که پاسخ‌ها به آن‌ها نیاز دارند:

</div>

```python
import numpy as np
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/keyvanmahmoudi/Data_Science_Course/main/data/"
df = pd.read_csv(DATA_URL + "penguins.csv")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۱

میانگین طول باله (`flipper_length_mm`) هر گونه را جداگانه با فیلتر کردن حساب کنید. (جلسهٔ ۴ راه کوتاه‌تری با `groupby` یاد می‌گیریم.)

</div>

```python
for species in df["species"].unique():
    mean_flipper = df.loc[df["species"] == species, "flipper_length_mm"].mean()
    print(f"{species:10s} {mean_flipper:.1f} mm")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۲

سنگین‌ترین پنگوئن مادهٔ هر جزیره را پیدا کنید.

</div>

```python
females = df[df["sex"] == "FEMALE"]
heaviest = females.sort_values("body_mass_g", ascending=False).drop_duplicates("island")
heaviest[["island", "species", "body_mass_g"]].sort_values("island")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۳

ستونی با نام `bill_area` بسازید که حاصل‌ضرب طول و عمق منقار باشد و ۵ پنگوئن با بزرگ‌ترین مقدار آن را نشان دهید.

</div>

```python
df["bill_area"] = df["bill_length_mm"] * df["bill_depth_mm"]
df.nlargest(5, "bill_area")[["species", "island", "bill_length_mm", "bill_depth_mm", "bill_area"]]
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۴

چند درصد پنگوئن‌های جزیرهٔ Biscoe وزنی بیش از ۵ کیلوگرم دارند؟

</div>

```python
biscoe = df[df["island"] == "Biscoe"]
share = (biscoe["body_mass_g"] > 5000).mean() * 100
print(f"{share:.1f}% of Biscoe penguins weigh more than 5 kg")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۵

با `pd.cut`، طول منقار را به سه دستهٔ «کوتاه»، «متوسط» و «بلند» تقسیم کنید (مرزها را خودتان با توجه به `describe` انتخاب کنید) و با `crosstab` رابطهٔ آن را با گونه ببینید.

</div>

```python
print(df["bill_length_mm"].describe().round(1))
df["bill_class"] = pd.cut(df["bill_length_mm"], bins=[0, 40, 48, np.inf],
                          labels=["short", "medium", "long"])
pd.crosstab(df["species"], df["bill_class"])
```
