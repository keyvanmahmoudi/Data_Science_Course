### آماده‌سازی

کتابخانه‌ها و داده‌هایی که پاسخ‌ها به آن‌ها نیاز دارند:

</div>

```python
import numpy as np
import pandas as pd
from scipy import stats

rng = np.random.default_rng(1405)
DATA_URL = "https://raw.githubusercontent.com/keyvanmahmoudi/Data_Science_Course/main/data/"
titanic = pd.read_csv(DATA_URL + "titanic.csv")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۱

با شبیه‌سازی، احتمال اینکه در ۱۰ پرتاب سکه حداقل ۳ شیر **پشت سر هم** بیاید را تخمین بزنید.

</div>

```python
flips = rng.integers(0, 2, size=(200_000, 10))       # 1 = heads
# a run of 3 heads ends at position i if flips i-2, i-1 and i are all heads
run3 = (flips[:, :-2] & flips[:, 1:-1] & flips[:, 2:]).any(axis=1)
print(f"P(at least 3 heads in a row in 10 flips) ≈ {run3.mean():.3f}   (exact: 520/1024 = {520 / 1024:.3f})")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۲

در مثال آزمایش تشخیصی، اگر نتیجهٔ آزمایش **دوم** (مستقل از اولی) هم مثبت شود، احتمال بیماری چقدر می‌شود؟ (راهنما: پسین مرحلهٔ اول، پیشین مرحلهٔ دوم است.)

</div>

```python
prior = 0.01
sensitivity, false_positive = 0.95, 0.05

def update(prior):
    return sensitivity * prior / (sensitivity * prior + false_positive * (1 - prior))

after_first = update(prior)
after_second = update(after_first)
print(f"after one positive test : {after_first:.3f}")
print(f"after two positive tests: {after_second:.3f}")
```

### تمرین ۳

یک فروشگاه به‌طور میانگین روزی ۲ مرجوعی دارد. احتمال اینکه در یک روز بیش از ۵ مرجوعی داشته باشد چقدر است؟ احتمال اینکه در یک هفتهٔ هفت‌روزه بیش از ۲۰ مرجوعی داشته باشد چطور؟

</div>

```python
day = stats.poisson(mu=2)
week = stats.poisson(mu=2 * 7)          # sum of independent Poisson counts is Poisson
print(f"P(more than 5 returns in a day)   = {day.sf(5):.4f}")
print(f"P(more than 20 returns in a week) = {week.sf(20):.4f}")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۴

نمرهٔ یک آزمون نرمال با میانگین ۱۲ و انحراف معیار ۳ است. برای قرار گرفتن در ۱۰٪ برتر، حداقل چه نمره‌ای لازم است؟

</div>

```python
score = stats.norm(loc=12, scale=3)
print(f"minimum score for the top 10%: {score.ppf(0.90):.2f}")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۵

در دادهٔ تایتانیک، $P(\text{survived} \mid \text{pclass} = 1)$ و $P(\text{pclass} = 1 \mid \text{survived})$ را حساب کنید و توضیح دهید چرا با هم برابر نیستند.

</div>

```python
p_surv_given_first = titanic.loc[titanic["pclass"] == 1, "survived"].mean()
p_first_given_surv = (titanic.loc[titanic["survived"] == 1, "pclass"] == 1).mean()
print(f"P(survived | pclass = 1) = {p_surv_given_first:.3f}")
print(f"P(pclass = 1 | survived) = {p_first_given_surv:.3f}")
```
