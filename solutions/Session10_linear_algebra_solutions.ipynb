### آماده‌سازی

کتابخانه‌ها و داده‌هایی که پاسخ‌ها به آن‌ها نیاز دارند:

</div>

```python
import numpy as np
import pandas as pd

np.set_printoptions(precision=3, suppress=True)
DATA_URL = "https://raw.githubusercontent.com/keyvanmahmoudi/Data_Science_Course/main/data/"
penguins = pd.read_csv(DATA_URL + "penguins.csv").dropna()

def cosine_similarity(a, b):
    return a @ b / (np.linalg.norm(a) * np.linalg.norm(b))
```

### تمرین ۱

شباهت کسینوسی بردار `[1, 2, 3]` با `[2, 4, 6]` و با `[-1, -2, -3]` را حساب کنید و توضیح دهید.

</div>

```python
v = np.array([1, 2, 3])
print("with [2, 4, 6]   :", cosine_similarity(v, np.array([2, 4, 6])))
print("with [-1, -2, -3]:", cosine_similarity(v, np.array([-1, -2, -3])))
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

`[2, 4, 6]` دو برابر `v` است: هم‌جهت، پس شباهت ۱. `[-1, -2, -3]` دقیقاً در خلاف جهت است: شباهت ۱-. طول بردار اثری ندارد.

</div>

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۲

ماتریس دوران ۹۰ درجه را بنویسید و نشان دهید که چهار بار اعمال آن، ماتریس همانی می‌دهد.

</div>

```python
R = np.array([[0, -1], [1, 0]])                    # rotate 90 degrees counter-clockwise
print(np.linalg.matrix_power(R, 4))
print("equals identity:", np.array_equal(np.linalg.matrix_power(R, 4), np.eye(2)))
```

### تمرین ۳

یک دستگاه سه معادله و سه مجهول بنویسید و با `np.linalg.solve` حل کنید. ضرایب را طوری انتخاب کنید که دستگاه جواب نداشته باشد؛ چه خطایی می‌گیرید؟

</div>

```python
A = np.array([[2, 1, -1], [1, 3, 2], [1, 0, 1]])
y = np.array([3, 13, 4])
print("solution:", np.linalg.solve(A, y))

singular = np.array([[1, 2, 3], [2, 4, 6], [1, 0, 1]])     # row 2 = 2 * row 1
try:
    np.linalg.solve(singular, y)
except np.linalg.LinAlgError as err:
    print("LinAlgError:", err, "| det =", round(np.linalg.det(singular), 10))
```

### تمرین ۴

با کمترین مربعات، `body_mass_g` را از **هر سه** ویژگی عددی دیگر پنگوئن (طول باله، طول و عمق منقار) پیش‌بینی کنید. $R^2$ چقدر بهتر شد؟

</div>

```python
y = penguins["body_mass_g"].values

def r_squared(features):
    X = np.column_stack([np.ones(len(penguins))] + [penguins[f].values for f in features])
    beta, *_ = np.linalg.lstsq(X, y, rcond=None)
    pred = X @ beta
    return 1 - ((y - pred) ** 2).sum() / ((y - y.mean()) ** 2).sum()

print("flipper only        R^2 =", round(r_squared(["flipper_length_mm"]), 3))
print("all three predictors R^2 =", round(r_squared(["flipper_length_mm", "bill_length_mm", "bill_depth_mm"]), 3))
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

بهبود تقریباً صفر است. طول باله به‌تنهایی بیشتر اطلاعات لازم را دارد و دو ویژگی منقار با آن همبسته‌اند، پس اطلاعات تازهٔ چندانی اضافه نمی‌کنند. (اگر **گونه** را هم به مدل اضافه کنیم، $R^2$ بیشتر بالا می‌رود؛ این را در ترم یادگیری ماشین می‌بینیم.)

</div>

### تمرین ۵

PCA را بدون استانداردسازی (فقط با کم کردن میانگین) تکرار کنید. چرا تقریباً همهٔ واریانس در مؤلفهٔ اول قرار می‌گیرد؟ (راهنما: واحد `body_mass_g`)

</div>

```python
features = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
data = penguins[features].values
centred = data - data.mean(axis=0)
C = centred.T @ centred / (len(centred) - 1)
eigvals = np.linalg.eigvalsh(C)[::-1]
print("explained variance ratio (no scaling):", (eigvals / eigvals.sum()).round(4))
print("variance of each column:", pd.Series(data.var(axis=0, ddof=1), index=features).round(1).to_dict())
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

واریانس `body_mass_g` (بر حسب گرم) ده‌ها هزار برابر بقیهٔ ستون‌هاست (میلی‌متر). بدون استانداردسازی، مؤلفهٔ اول تقریباً همان ستون وزن است و بقیهٔ ویژگی‌ها عملاً نادیده گرفته می‌شوند. PCA به **واحد** حساس است.
