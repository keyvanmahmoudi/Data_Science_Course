## ۱. Series: یک ستون با برچسب

`Series` یک آرایه‌ی یک‌بعدی است که هر عنصرش یک **برچسب (index)** دارد.

</div>

```python
import numpy as np
import pandas as pd

population = pd.Series(
    [9.0, 3.4, 2.2, 1.8, 1.6],
    index=["Tehran", "Mashhad", "Isfahan", "Karaj", "Shiraz"],
    name="population_millions",
)
population
```

```python
print(population["Isfahan"])           # access by label
print(population.iloc[0])              # access by position
print(population[population > 2])      # boolean filtering, just like NumPy
print("mean:", round(population.mean(), 2))
```

## ۶. انتخاب سطر و ستون با `loc` و `iloc`

- `loc` بر اساس **برچسب** کار می‌کند و انتهای بازه را **شامل** می‌شود.
- `iloc` بر اساس **موقعیت عددی** کار می‌کند و مثل پایتون، انتهای بازه را شامل **نمی‌شود**.

شکل کلی هر دو: `df.loc[سطرها, ستون‌ها]`

</div>

```python
df.loc[0:3, ["species", "bill_length_mm"]]     # rows labelled 0..3 (4 rows)
```

```python
df.iloc[0:3, 0:4]                              # positions 0..2 (3 rows), first 4 columns
```

همان اندیس‌گذاری بولی NumPy، این بار روی جدول. برای ترکیب شرط‌ها از `&` و `|` استفاده کنید و هر شرط را در پرانتز بگذارید.

## ۹. ساختن و تغییر ستون

عملیات روی ستون‌ها، مثل NumPy، برداری است:

</div>

```python
df["body_mass_kg"] = df["body_mass_g"] / 1000
df["bill_ratio"] = df["bill_length_mm"] / df["bill_depth_mm"]
df["size_class"] = np.where(df["body_mass_g"] >= 4500, "large", "small")
df[["species", "body_mass_kg", "bill_ratio", "size_class"]].head()
```

<div dir="rtl">

برای دسته‌بندی یک ستون عددی در چند بازه، `pd.cut` مناسب است:

</div>

```python
df["mass_band"] = pd.cut(
    df["body_mass_g"],
    bins=[0, 3500, 4500, 5500, np.inf],
    labels=["light", "medium", "heavy", "very heavy"],
)
df["mass_band"].value_counts().sort_index()
```

## ۱۰. `map` و `apply`

- `Series.map(dict)`: جایگزینی مقادیر بر اساس یک دیکشنری
- `Series.map(func)` یا `Series.apply(func)`: اعمال یک تابع روی تک‌تک مقادیر
- `DataFrame.apply(func, axis=1)`: اعمال تابع روی هر **سطر**

> **هشدار کارایی:** `apply` در واقع یک حلقه‌ی پایتونی است. هر جا عملیات برداری وجود دارد (مثل بخش قبل)، آن را ترجیح دهید و `apply` را برای منطق‌های پیچیده نگه دارید.

</div>

```python
sex_map = {"MALE": "M", "FEMALE": "F"}
df["sex_short"] = df["sex"].map(sex_map)

def describe_penguin(row):
    if pd.isna(row["body_mass_g"]):
        return "unknown"
    return f"{row['species']} from {row['island']}, {row['body_mass_g'] / 1000:.1f} kg"

df["description"] = df.apply(describe_penguin, axis=1)
df[["sex", "sex_short", "description"]].head()
```

## ۱۱. تغییر نام، حذف ستون و ذخیره‌سازی

- به زنجیره کردن متدها با پرانتز در کد بالا دقت کنید؛ این سبک (Method Chaining) کد را خوانا نگه می‌دارد.

</div>
